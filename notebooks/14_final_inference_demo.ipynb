```python
# TensorFlow
import tensorflow as tf

# Numerical operations
import numpy as np

# Visualization
import matplotlib.pyplot as plt

# OpenCV
import cv2

# File handling
import os

# Load trained model
from tensorflow.keras.models import load_model

# Image preprocessing
from tensorflow.keras.preprocessing import image
```

```python
# Load trained EfficientNet model
model = load_model(
    "../models/efficientnet_papsmear.h5"
)

# Display model summary
model.summary()
```

```python
# Define Pap smear classes
class_names = [

    "Dyskeratotic",

    "Koilocytotic",

    "Metaplastic",

    "Parabasal",

    "Superficial-Intermediate"
]
```

```python
# IMPORTANT:
#Replace:

#sample.bmp
# Define unseen image path
img_path = "../dataset/Parabasal/sample.bmp"
```

```python
# Load image
img = image.load_img(

    img_path,

    target_size=(224,224)
)

# Convert image to array
img_array = image.img_to_array(img)

# Expand dimensions
img_array = np.expand_dims(
    img_array,
    axis=0
)

# Normalize image
img_array = img_array / 255.0
```

```python
# Display image
plt.imshow(img)

# Add title
plt.title("Input Pap Smear Image")

# Remove axis
plt.axis('off')

# Show image
plt.show()
```

```python
# Predict image probabilities
predictions = model.predict(img_array)

# Retrieve predicted class index
predicted_class = np.argmax(
    predictions
)

# Retrieve confidence score
confidence = np.max(
    predictions
)

# Retrieve class name
predicted_label = class_names[
    predicted_class
]
```

```python
# Display predicted class
print("Predicted Class:", predicted_label)

# Display confidence score
print("Confidence:", confidence)
```

```python
# Loop through class probabilities
for i, probability in enumerate(predictions[0]):

    print(
        class_names[i],
        ":",
        probability
    )
```

```python
# Create output folder
os.makedirs(
    "../outputs/inference_results",
    exist_ok=True
)

# Save prediction result
with open(

    "../outputs/inference_results/result.txt",

    "w"

) as file:

    file.write(
        f"Predicted Class: {predicted_label}\n"
    )

    file.write(
        f"Confidence: {confidence}"
    )
```
