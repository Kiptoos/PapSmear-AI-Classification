```python
# Import Required Libraries
```

```python
# TensorFlow for deep learning
import tensorflow as tf

# Numerical computations
import numpy as np

# Data handling
import pandas as pd

# Visualization
import matplotlib.pyplot as plt
import seaborn as sns

# File and folder handling
import os

# Image preprocessing utilities
from tensorflow.keras.preprocessing.image import ImageDataGenerator

# Pretrained ResNet50 model
from tensorflow.keras.applications import ResNet50

# Neural network layers and model creation
from tensorflow.keras import layers, models

# Evaluation metrics
from sklearn.metrics import classification_report
from sklearn.metrics import confusion_matrix
```

```python
# Verify TensorFlow Version
```

```python
# Print TensorFlow version
print("TensorFlow Version:", tf.__version__)
```

```python
# Define Dataset Path
```

```python
# Path to dataset directory
dataset_path = "../dataset"

# Display available class folders
print(os.listdir(dataset_path))
```

```python
# Count Images Per Class
```

```python
# Loop through dataset folders
for folder in os.listdir(dataset_path):

    # Full path
    path = os.path.join(dataset_path, folder)

    # Check if directory exists
    if os.path.isdir(path):

        # Print class name and image count
        print(folder, len(os.listdir(path)))
```

```python
# Load and Preprocess Images
```

```python
# Define image dimensions
IMG_SIZE = 224

# Define batch size
BATCH_SIZE = 32

# Create image generator
datagen = ImageDataGenerator(

    # Normalize pixel values
    rescale=1./255,

    # Validation split
    validation_split=0.2
)

# Load training images
train_data = datagen.flow_from_directory(
    dataset_path,
    target_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    subset='training'
)

# Load validation images
val_data = datagen.flow_from_directory(
    dataset_path,
    target_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    subset='validation'
)
```

```python
# Display Class Labels
```

```python
# Print class labels
print(train_data.class_indices)
```

```python
# Visualize Sample Images
```

```python
# Retrieve one batch of images
images, labels = next(train_data)

# Create display figure
plt.figure(figsize=(12,12))

# Display first 9 images
for i in range(9):

    plt.subplot(3,3,i+1)
    plt.imshow(images[i])
    plt.axis('off')

# Show images
plt.show()
```

```python
# Build ResNet50 Model
```

```python
# Load pretrained ResNet50 model
base_model = ResNet50(
    weights='imagenet',
    include_top=False,
    input_shape=(224,224,3)
)
```

```python
# Freeze pretrained layers
base_model.trainable = False

# Build classification model
model = models.Sequential([

    # ResNet50 feature extractor
    base_model,

    # Convert feature maps to vectors
    layers.GlobalAveragePooling2D(),

    # Dense layer
    layers.Dense(128, activation='relu'),

    # Reduce overfitting
    layers.Dropout(0.5),

    # Output layer for 5 classes
    layers.Dense(5, activation='softmax')
])
```

```python
# Compile Model
```

```python
# Configure model training
model.compile(

    # Optimization algorithm
    optimizer='adam',

    # Multi-class classification loss
    loss='categorical_crossentropy',

    # Accuracy metric
    metrics=['accuracy']
)
```

```python
 # View Model Summary
```

```python
# Display model architecture
model.summary()
```

```python
# Train Model
```

```python
# Train the ResNet50 model
history = model.fit(

    # Training dataset
    train_data,

    # Validation dataset
    validation_data=val_data,

    # Number of epochs
    epochs=10
)
```

```python
# Plot Accuracy Graph
```

```python
# Plot training accuracy
plt.plot(
    history.history['accuracy'],
    label='Training Accuracy'
)

# Plot validation accuracy
plt.plot(
    history.history['val_accuracy'],
    label='Validation Accuracy'
)

# Labels
plt.xlabel('Epoch')
plt.ylabel('Accuracy')

# Show legend
plt.legend()

# Display graph
plt.show()
```

```python
# Plot Loss Graph
```

```python
# Plot training loss
plt.plot(
    history.history['loss'],
    label='Training Loss'
)

# Plot validation loss
plt.plot(
    history.history['val_loss'],
    label='Validation Loss'
)

# Labels
plt.xlabel('Epoch')
plt.ylabel('Loss')

# Show legend
plt.legend()

# Display graph
plt.show()
```

```python
# Evaluate Model
```

```python
# Evaluate model performance
loss, accuracy = model.evaluate(val_data)

# Print validation accuracy
print("Validation Accuracy:", accuracy)
```

```python
# Generate Predictions
```

```python
# Generate predictions
predictions = model.predict(val_data)

# Convert probabilities into class labels
predicted_classes = np.argmax(predictions, axis=1)

# Actual class labels
true_classes = val_data.classes

# Retrieve class names
class_labels = list(val_data.class_indices.keys())
```

```python
# Classification Report
```

```python
# Print precision, recall and F1-score
print(classification_report(
    true_classes,
    predicted_classes,
    target_names=class_labels
))
```

```python
# Confusion Matrix
```

```python
# Generate confusion matrix
cm = confusion_matrix(
    true_classes,
    predicted_classes
)

# Create visualization figure
plt.figure(figsize=(8,6))

# Display heatmap
sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=class_labels,
    yticklabels=class_labels
)

# Labels
plt.xlabel('Predicted')
plt.ylabel('Actual')

# Title
plt.title('Confusion Matrix')

# Display heatmap
plt.show()
```
