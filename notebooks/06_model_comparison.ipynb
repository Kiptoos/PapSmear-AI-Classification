```python
# Data analysis
import pandas as pd

# Visualization
import matplotlib.pyplot as plt

# Numerical operations
import numpy as np
```

```python
# IMPORTANT:
# Replace these example values with YOUR actual results after experiments.
    
# Create dictionary containing model results
results = {

    # Model names
    'Model': [
        'EfficientNetB0',
        'ResNet50',
        'DenseNet169'
    ],

    # Validation accuracy
    'Accuracy': [
        0.91,
        0.88,
        0.89
    ],

    # Precision scores
    'Precision': [
        0.90,
        0.87,
        0.88
    ],

    # Recall scores
    'Recall': [
        0.91,
        0.88,
        0.89
    ],

    # F1-score values
    'F1-Score': [
        0.90,
        0.87,
        0.88
    ]
}
```

```python
# Convert dictionary into DataFrame
df = pd.DataFrame(results)

# Display table
print(df)
```

```python
# Create figure
plt.figure(figsize=(8,5))

# Plot model accuracy
plt.bar(
    df['Model'],
    df['Accuracy']
)

# Axis labels
plt.xlabel('Model')
plt.ylabel('Accuracy')

# Graph title
plt.title('Model Accuracy Comparison')

# Display graph
plt.show()
```

```python
# Create figure
plt.figure(figsize=(8,5))

# Plot precision scores
plt.bar(
    df['Model'],
    df['Precision']
)

# Axis labels
plt.xlabel('Model')
plt.ylabel('Precision')

# Graph title
plt.title('Model Precision Comparison')

# Display graph
plt.show()
```

```python
# Create figure
plt.figure(figsize=(8,5))

# Plot recall scores
plt.bar(
    df['Model'],
    df['Recall']
)

# Axis labels
plt.xlabel('Model')
plt.ylabel('Recall')

# Graph title
plt.title('Model Recall Comparison')

# Display graph
plt.show()
```

```python
# Create figure
plt.figure(figsize=(8,5))

# Plot F1-score
plt.bar(
    df['Model'],
    df['F1-Score']
)

# Axis labels
plt.xlabel('Model')
plt.ylabel('F1-Score')

# Graph title
plt.title('Model F1-Score Comparison')

# Display graph
plt.show()
```

```python
# Save comparison results as CSV
df.to_csv(
    "../outputs/model_comparison.csv",
    index=False
)
```
