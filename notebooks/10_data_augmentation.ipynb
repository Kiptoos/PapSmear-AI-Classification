```python
# TensorFlow
import tensorflow as tf

# Numerical operations
import numpy as np

# Visualization
import matplotlib.pyplot as plt

# File handling
import os

# Image preprocessing
from tensorflow.keras.preprocessing.image import ImageDataGenerator
```

```python
# Dataset location
dataset_path = "../dataset"

# Display dataset folders
print(os.listdir(dataset_path))
```

```python
# Define image dimensions
IMG_SIZE = 224

# Define batch size
BATCH_SIZE = 32

# Create augmentation pipeline
datagen = ImageDataGenerator(

    # Normalize images
    rescale=1./255,

    # Random rotation
    rotation_range=20,

    # Horizontal shifting
    width_shift_range=0.2,

    # Vertical shifting
    height_shift_range=0.2,

    # Zoom augmentation
    zoom_range=0.2,

    # Shearing transformation
    shear_range=0.2,

    # Horizontal flip
    horizontal_flip=True,

    # Validation split
    validation_split=0.2
)
```

```python
# Load augmented training dataset
train_data = datagen.flow_from_directory(

    dataset_path,

    target_size=(IMG_SIZE, IMG_SIZE),

    batch_size=BATCH_SIZE,

    class_mode='categorical',

    subset='training'
)
```

```python
# Load validation dataset
val_data = datagen.flow_from_directory(

    dataset_path,

    target_size=(IMG_SIZE, IMG_SIZE),

    batch_size=BATCH_SIZE,

    class_mode='categorical',

    subset='validation'
)
```

```python
# Retrieve augmented batch
images, labels = next(train_data)

# Create figure
plt.figure(figsize=(12,12))

# Display augmented images
for i in range(9):

    plt.subplot(3,3,i+1)

    plt.imshow(images[i])

    plt.axis('off')

# Display images
plt.show()
```

```python
# Create output directory
os.makedirs(
    "../outputs/augmented_samples",
    exist_ok=True
)

# Save first 5 augmented images
for i in range(5):

    plt.imsave(
        f"../outputs/augmented_samples/sample_{i}.png",
        images[i]
    )
```
