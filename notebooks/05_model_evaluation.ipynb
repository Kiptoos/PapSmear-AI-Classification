```python
# TensorFlow
import tensorflow as tf

# Numerical operations
import numpy as np

# Visualization
import matplotlib.pyplot as plt

# Image generator
from tensorflow.keras.preprocessing.image import ImageDataGenerator

# Load saved model
from tensorflow.keras.models import load_model

# EfficientNet preprocessing
from tensorflow.keras.applications.efficientnet import preprocess_input
```

```python
from tensorflow.keras.models import load_model

# Load model without optimizer
model = load_model(

    "../models/efficientnet_papsmear.keras",

    compile=False
)
```

```python
# Define image size
IMG_SIZE = 224

# Batch size
BATCH_SIZE = 32

# Create image generator
datagen = ImageDataGenerator(

    # EfficientNet preprocessing
    preprocessing_function=preprocess_input,

    # Validation split
    validation_split=0.2
)

# Load ONLY validation dataset
val_data = datagen.flow_from_directory(

    # Dataset path
    "../dataset",

    # Resize images
    target_size=(IMG_SIZE, IMG_SIZE),

    # Batch size
    batch_size=BATCH_SIZE,

    # Multi-class labels
    class_mode='categorical',

    # Use validation subset only
    subset='validation',

    # Keep order fixed for evaluation
    shuffle=False
)
```

```python
# Predict validation images
predictions = model.predict(val_data)

# Convert probabilities to class labels
predicted_classes = np.argmax(
    predictions,
    axis=1
)

# Retrieve actual labels
true_classes = val_data.classes

# Retrieve class names
class_labels = list(
    val_data.class_indices.keys()
)
```
