```python
# Data analysis
import pandas as pd

# Numerical operations
import numpy as np
```

```python
# Create experiment results dictionary
results = {

    # CNN model names
    "Model": [
        "EfficientNetB0",
        "ResNet50",
        "DenseNet169"
    ],

    # Accuracy values
    "Accuracy": [
        0.91,
        0.88,
        0.89
    ],

    # Precision values
    "Precision": [
        0.90,
        0.87,
        0.88
    ],

    # Recall values
    "Recall": [
        0.91,
        0.88,
        0.89
    ],

    # F1-score values
    "F1-Score": [
        0.90,
        0.87,
        0.88
    ],

    # AUC scores
    "AUC": [
        0.95,
        0.92,
        0.93
    ]
}
```

```python
# Convert dictionary into DataFrame
df = pd.DataFrame(results)

# Display results table
print(df)
```

```python
# Round results to 2 decimal places
df = df.round(2)

# Display rounded table
print(df)
```

```python
# Save results table
df.to_csv(

    "../outputs/research_results.csv",

    index=False
)
```

```python
# Save results as Excel
df.to_excel(

    "../outputs/research_results.xlsx",

    index=False
)
```

```python
# Convert DataFrame into LaTeX table
latex_table = df.to_latex(index=False)

# Display LaTeX code
print(latex_table)
```

```python
# Save LaTeX table
with open(

    "../outputs/results_table.tex",

    "w"

) as file:

    file.write(latex_table)
```
